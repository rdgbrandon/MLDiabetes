```python
import pandas as pd
```

```python
df = pd.read_csv('datasetDia.csv')
df = df.rename(columns={'Diabetes_012': 'diabetes'})
```

```python
df['diabetes'].value_counts()
```

```python
df['diabetes'] = df['diabetes'].replace(2, 1)
```

```python
df_zeros = df[df['diabetes'] == 0]
df_ones = df[df['diabetes'] == 1]
df_zeros_downsampled = df_zeros.sample(n=len(df_ones), random_state=42)
```

```python
df = pd.concat([df_zeros_downsampled, df_ones])
df = df.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
print(df['diabetes'].value_counts())
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
from sklearn.metrics import confusion_matrix
```

```python
from sklearn.metrics import classification_report
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop(columns=['diabetes'])
```

```python
y = df.iloc[:,0]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=11,test_size=0.2)
```

```python
scaler = MinMaxScaler(feature_range=(0,1))
```

```python
X_train = scaler.fit_transform(X_train)
```

```python
X_test = scaler.fit_transform(X_test)
```

```python
knn = KNeighborsClassifier(n_neighbors=8)
```

```python
knn.fit(X_train, y_train)
```

```python
y_pred = knn.predict(X_test)
```

```python
knn.score(X_test,y_test)
```

```python
cm = confusion_matrix(y_test,y_pred)
```

```python
cm
```

```python
print(classification_report(y_test, y_pred))
```

```python
from sklearn.inspection import permutation_importance
```

```python
results = permutation_importance(knn, X_test, y_test, n_repeats=10, random_state=42)
```

```python
feature = pd.DataFrame(results.importances_mean, index=X.columns, columns=['importance'])
```

```python
print(feature)
```
